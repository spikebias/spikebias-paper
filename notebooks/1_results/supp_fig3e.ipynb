```python
# SETUP PACKAGES
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib
from matplotlib import pyplot as plt
import spikeinterface as si
import spikeinterface.preprocessing as spre
warnings.simplefilter(action="ignore")

# SET PATHS
PROJ_PATH = "/home/steeve/steeve/epfl/code/spikebias-paper/"
os.chdir(PROJ_PATH)
from src.nodes.validation.burst_spikes import find_bursts_isi, mean_waveforms_by_isi_range
import seaborn as sns
from scipy.stats import wilcoxon
REC_PATH = "dataset/00_raw/recording_npx_evoked"
GT_PATH = "dataset/00_raw/ground_truth_npx_evoked"

# PARAMETERS
FREQ_MIN, FREQ_MAX = 300, 3000      # band-pass filter
MIN_SPIKES = 2                      # minimum number of spikes in a burst
ISI_THRESH_SEC = 0.021              # maximum inter-spike interval within a burst
MIN_SPIKES_SEQUENCE = 2             # minimum number of spikes in a sequence within an ISI range
NUM_SPIKES = 2                      # first spikes of each sequence
MIN_DURATION = 0.02                 # minimum duration of a sequence
MIN_TROUGH_AMPLITUDE = -7.5         # minimum trough amplitude
WINDOW_MS = 2.5                     # waveform duration
MIN_ISIS = [0.021, 0.061, 0.101, 0.141]         # ISI ranges
MAX_ISIS = [0.061, 0.101, 0.141, 0.181]
ISI_LABELS = ["21-61", "61-101", "101-141", "141-181"]

# FIGURE SETTINGS
plt.rcParams["font.family"] = "Arial"
plt.rcParams["font.size"] = 6  # 5-7 with Nature neuroscience as reference
plt.rcParams["lines.linewidth"] = 0.5
plt.rcParams["axes.linewidth"] = 0.5
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
plt.rcParams["xtick.major.width"] = 0.5
plt.rcParams["ytick.major.width"] = 0.5
plt.rcParams["xtick.major.size"] = 3.5 * 1.1
plt.rcParams["ytick.major.size"] = 3.5 * 1.1
savefig_cfg = {"transparent": True, "dpi": 300}
```

## Load the recording and the ground truth, find the non-bursting units

```python
%%time

# recording: preprocessed lazily
Recording = si.load_extractor(REC_PATH)
Recording = spre.astype(Recording, "int16")
Recording = spre.bandpass_filter(Recording, freq_min=FREQ_MIN, freq_max=FREQ_MAX)
Recording = spre.common_reference(Recording, reference="global", operator="median")
Recording = spre.whiten(Recording, dtype="float32")
sfreq = Recording.get_sampling_frequency()
window_frames = int(WINDOW_MS / (1 / sfreq) / 1000)

# ground truth: non-bursting units
Gtruth = si.load_extractor(GT_PATH)
bursts = {u: find_bursts_isi(Gtruth.get_unit_spike_train(u) / sfreq, isi_threshold=ISI_THRESH_SEC, min_spikes=MIN_SPIKES) for u in Gtruth.unit_ids}
non_bursting_units = [u for u in Gtruth.unit_ids if len(bursts[u]) == 0]
print("number of non-bursting units:", len(non_bursting_units))
```

## Trough amplitudes of the first and second spikes for each ISI range (about 30 minutes)

```python
%%time

rows = []
for min_isi, max_isi, label in zip(MIN_ISIS, MAX_ISIS, ISI_LABELS):
    mean_waves = mean_waveforms_by_isi_range(Recording, Gtruth, non_bursting_units, sfreq, min_isi, max_isi, MIN_SPIKES_SEQUENCE, window_frames, NUM_SPIKES, min_duration=MIN_DURATION, min_amplitude=MIN_TROUGH_AMPLITUDE)
    for unit, waves in mean_waves.items():
        for s, name in enumerate(["1st", "2nd"]):
            rows.append({"isi_range_ms": label, "unit": unit, "spike_rank": name, "trough_amplitude_uV": waves[s].min()})
    print(label, "ms:", len(mean_waves), "units")
df = pd.DataFrame(rows)
```

## Plot, same as the figure d

```python
def stars(p):
    return "***" if p < 0.001 else "**" if p < 0.01 else "*" if p < 0.05 else "ns"


def test_isi_range(d):
    """One-sided paired Wilcoxon signed-rank test that the trough of the 2nd spike is shallower than the 1st's"""
    first = d[d["spike_rank"] == "1st"].set_index("unit")["trough_amplitude_uV"]
    second = d[d["spike_rank"] == "2nd"].set_index("unit")["trough_amplitude_uV"].loc[first.index]
    z, p = wilcoxon(first, y=second, zero_method="wilcox", correction=False, alternative="less")
    return len(first), z, p


from src.nodes.validation.burst_plots import plot_trough_amplitudes_non_bursting

def plot_source_data(df: pd.DataFrame):
    stats = pd.DataFrame([dict(isi_range_ms=l, n_units=n, wilcoxon_signed_rank_z=z, p_one_sided=p, significance=stars(p)) for l in ISI_LABELS for n, z, p in [test_isi_range(df[df["isi_range_ms"] == l])]])
    return plot_trough_amplitudes_non_bursting((df, stats))
```

## Plot and save the figure and the source data

```python
fig = plot_source_data(df)
os.makedirs("figures/source_data/supp_fig3e", exist_ok=True)
plt.savefig("figures/source_data/supp_fig3e/supp_fig3e.svg", transparent=True)   # the .pdf is written next to the .svg

# statistics of each panel
stats = pd.DataFrame([dict(isi_range_ms=l, n_units=n, wilcoxon_signed_rank_z=z, p_one_sided=p, significance=stars(p)) for l in ISI_LABELS for n, z, p in [test_isi_range(df[df["isi_range_ms"] == l])]])
from statsmodels.stats.multitest import multipletests
stats.insert(stats.columns.get_loc("p_one_sided") + 1, "p_holm", multipletests(stats["p_one_sided"], method="holm")[1])     # Holm correction for the ISI ranges of the panel
display(stats)
with pd.ExcelWriter("figures/source_data/supp_fig3e/supp_fig3e.xlsx") as writer:
    df.to_excel(writer, sheet_name="trough_amplitudes", index=False)
    stats.to_excel(writer, sheet_name="statistics", index=False)
```
